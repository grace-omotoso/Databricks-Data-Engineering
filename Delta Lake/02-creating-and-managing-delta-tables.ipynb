### Three Ways to Create Delta Table

```python
%sql
use catalog workspace;
use schema default;
```

#### Using CREATE TABLE STATEMENT

```python
%sql
CREATE OR REPLACE TABLE products( 
    product_id INT COMMENT 'Unique Product Identifier',
    name STRING COMMENT 'Product Name',
    category STRING COMMENT 'Product Category',
    price DOUBLE COMMENT 'Product Price',
    is_stock BOOLEAN COMMENT 'Is product available'

) USING DELTA
COMMENT 'Product Catalogue - create with explicit DDL';
```

```python
%sql
INSERT into products VALUES(
    1, 'Wireless Headphone', 'Electronics', 100.0, true
),
(
    2, 'Standing Desk', 'Furniture', 200.0, false            
),
(
    3, 'Eco-Friendly T-Shirt', 'Clothing', 25.0, true
),(
    4, 'Smartwatch', 'Electronics', 150.0, true
),
(
    5, 'Organic Cotton Pants', 'Clothing', 30.0, true
),
(
    6, 'Eco-Friendly Shoes', 'Footwear', 50.0, true
)
```

#### Using CTAS Statement

```python
%sql
CREATE OR REPLACE TABLE electronics
USING DELTA
COMMENT 'Electronics subset - Created with CTAS'
AS
SELECT * FROM products WHERE category = 'Electronics';
    
```

```python
%sql
select * from electronics;
```

```python
%sql
describe electronics;
```

#### Spark Data Frame

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DoubleType
schema = StructType([
  StructField("id", IntegerType(), True),
  StructField("name", StringType(), True),
  StructField("age", IntegerType(), True),
  StructField("salary", DoubleType(), True)
])

data = [(1, "Alice", 34, 100000.0),
        (2, "Bob", 45, 120000.0),
        (3, "Charlie", 27, 80000.0),
        (4, "David", 51, 150000.0),
        (5, "Eve", 39, 90000.0)]

df = spark.createDataFrame(data, schema)
df.write.format("delta").mode("overwrite").saveAsTable("workspace.default.employees")
```
